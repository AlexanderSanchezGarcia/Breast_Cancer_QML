```python
import pandas as pd, os, re

def build_path_map(base):
    """
    Construye un diccionario:
      SeriesUID → ruta local absoluta a la carpeta con DICOMs
    """
    meta = pd.read_csv(f"{base}/metadata.csv")
    path_map = {}
    for _, row in meta.iterrows():
        uid = row['Series UID']
        # File Location viene como './CBIS-DDSM/...' o 'CBIS-DDSM/...'
        loc = row['File Location'].lstrip('/')[0:2]
        full_path = os.path.join(base, loc)
        path_map[uid] = full_path
    return path_map, meta

def get_dicom_path(folder):
    """Devuelve el primer .dcm que encuentre en la carpeta"""
    for f in os.listdir(folder):
        if f.endswith('.dcm'):
            return os.path.join(folder, f)
    # A veces hay subcarpeta adicional
    for sub in os.listdir(folder):
        sub_path = os.path.join(folder, sub)
        if os.path.isdir(sub_path):
            for f in os.listdir(sub_path):
                if f.endswith('.dcm'):
                    return os.path.join(sub_path, f)
    return None
```

```python
def build_dataset(base, case_type='mass', split='train'):
    path_map, meta = build_path_map(base)
    
    csv_path = f"{base}/{case_type}_case_description_{split}_set.csv"
    df = pd.read_csv(csv_path)
    
    # Binarizar etiqueta
    df['label'] = (df['pathology'] == 'MALIGNANT').astype(int)
    
    records = []
    for _, row in df.iterrows():
        subject_id = row['image file path'].split('/')[0]
        
        # Buscar en metadata por Subject ID
        meta_full  = meta[meta['Subject ID'] == subject_id]
        meta_image = meta_full[meta_full['Series Description'].str.contains('full mammogram', case=False, na=False)]
        meta_mask  = meta_full[meta_full['Series Description'].str.contains('ROI mask', case=False, na=False)]
        
        if meta_image.empty or meta_mask.empty:
            continue  # caso incompleto, saltar
        
        img_uid  = meta_image.iloc[0]['Series UID']
        mask_uid = meta_mask.iloc[0]['Series UID']
        
        img_folder  = path_map.get(img_uid)
        mask_folder = path_map.get(mask_uid)
        
        if not img_folder or not mask_folder:
            continue
        
        img_dcm  = get_dicom_path(img_folder)
        mask_dcm = get_dicom_path(mask_folder)
        
        if img_dcm and mask_dcm:
            records.append({
                'patient_id':  subject_id,
                'img_path':    img_dcm,
                'mask_path':   mask_dcm,
                'label':       row['label'],
                'pathology':   row['pathology'],
                'side':        row.get('side', ''),
                'view':        row.get('image view', ''),
            })
            
    return pd.DataFrame(records)

        
```

```python
df_train = build_dataset(BASE, case_type='mass', split='train')
df_test  = build_dataset(BASE, case_type='mass', split='test')
print(f"Train: {len(df_train)} casos | Test: {len(df_test)} casos")
print(df_train[['patient_id','label','pathology']].value_counts('pathology'))
```
